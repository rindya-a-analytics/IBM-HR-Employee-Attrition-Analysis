```python
df["Attrition"].value_counts()
```

```python
df["Attrition"].value_counts(normalize=True) * 100
```

```python
import matplotlib.pyplot as plt
```

```python
df["Attrition"].value_counts().plot(kind="bar")

plt.title("Employee Attrition")
plt.xlabel("Attrition")
plt.ylabel("Number of Employees")
plt.show()
```

```python
df["OverTime"].value_counts()
```

```python
pd.crosstab(df["OverTime"], df["Attrition"])
```

```python
pd.crosstab(df["OverTime"], df["Attrition"], normalize="index") * 100
```

```python
pd.crosstab(
    df["OverTime"],
    df["Attrition"],
    normalize="index"
).mul(100).plot(kind="bar")

plt.title("Attrition Rate by Overtime")
plt.xlabel("Overtime")
plt.ylabel("Percentage (%)")
plt.legend(title="Attrition")
plt.xticks(rotation=0)
plt.show()
```

```python
pd.crosstab(
    df["JobSatisfaction"],
    df["Attrition"],
    normalize="index"
) * 100
```

```python
df.groupby("JobSatisfaction")["Attrition"].apply(
    lambda x: (x == "Yes").mean() * 100).plot(kind="bar")
plt.title("Attrition Rate by Job Satisfaction")
plt.xlabel("Job Satisfaction")
plt.ylabel("Percentage (%)")
plt.legend(title="Attrition")
plt.xticks(rotation=0)
plt.show()
```

```python
df.groupby("Attrition")["MonthlyIncome"].mean()
```

```python
df.boxplot(column="MonthlyIncome", by="Attrition")

plt.title("Monthly Income Distribution by Attrition")
plt.suptitle("")
plt.xlabel("Attrition")
plt.ylabel("Monthly Income")
plt.show()
```

```python
df.groupby("Attrition")["Age"].mean()
```

```python
df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=[17, 25, 35, 45, 55, 65],
    labels=["18-25", "26-35", "36-45", "46-55", "56-65"]
)
```

```python
df["AgeGroup"].value_counts().sort_index()
```

```python
age_attrition = df.groupby("AgeGroup")["Attrition"].apply(
    lambda x: (x == "Yes").mean() * 100
)

age_attrition
```

```python
age_attrition.plot(kind="bar")

plt.title("Attrition Rate by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Attrition Rate (%)")
plt.show()
```

```python
df.groupby("Attrition")["YearsAtCompany"].mean()
```

```python
df.groupby("Attrition")["YearsAtCompany"].mean().plot(kind="bar")

plt.title("Average Years at Company by Attrition")
plt.xlabel("Attrition")
plt.ylabel("Average Years at Company")
plt.show()
```
